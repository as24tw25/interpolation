```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d
from ipywidgets import interact, FloatText

# Функция загрузки и интерполяции данных
def load_and_interpolate_data(file_name):
    try:
        data = np.loadtxt(file_name, delimiter=',')
        if data.shape[1] != 2:
            raise ValueError("Файл должен содержать два столбца данных.")
        
        x = data[:, 0]
        y = data[:, 1]

        # Проводим интерполяцию
        interpolator = interp1d(x, y, kind='linear')

        return x, y, interpolator

    except Exception as e:
        print(f"Ошибка: {e}")
        return None, None, None

# Функция для вычисления значения в заданной точке
def interpolate_point(interpolator, point):
    try:
        result = interpolator(point)
        return result
    except Exception as e:
        print(f"Ошибка: {e}")
        return None

# Функция отрисовки графика
def plot_data(x, y, interpolator, point=None):
    plt.figure(figsize=(8, 6))
    plt.plot(x, y, 'o', label='Исходные данные')
    x_new = np.linspace(min(x), max(x), 300)
    y_new = interpolator(x_new)
    plt.plot(x_new, y_new, '-', label='Интерполяционная кривая')
    
    if point is not None:
        result = interpolate_point(interpolator, point)
        if result is not None:
            plt.scatter(point, result, color='red', label=f'Значение в точке {point}')
    
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title('График интерполированных данных')
    plt.legend()
    plt.show()

# Интерактивная функция
def interactive_plot(file_name, point):
    x, y, interpolator = load_and_interpolate_data(file_name)
    
    if x is not None and y is not None and interpolator is not None:
        plot_data(x, y, interpolator, point)

# Создание интерактивного виджета
point_widget = FloatText(value=5.0, description='Введите точку:')
interact(interactive_plot, file_name='data.csv', point=point_widget);
```
